```python
import pandas as pd
import matplotlib.pyplot as plt
from imblearn.over_sampling import ADASYN
from collections import Counter
```

```python
df = pd.read_csv('encoded_dataset.csv')
df
```

```python
X = df.drop('CLASS', axis=1)
y = df['CLASS']
```

```python
print("Original class distribution:", Counter(y))
```

### Apply ADASYN

```python
adasyn = ADASYN(random_state=42)
X_adasyn, y_adasyn = adasyn.fit_resample(X, y)
```

```python
print("ADASYN class distribution:", Counter(y_adasyn))
```

```python
X_adasyn_df = pd.DataFrame(X_adasyn, columns=X.columns)
y_adasyn_df = pd.Series(y_adasyn, name='CLASS')
```

```python
balanced_df = pd.concat([X_adasyn_df, y_adasyn_df], axis=1)
```

```python
balanced_df.to_csv('adasyn_balanced_dataset.csv', index=False)
```

```python
df = pd.read_csv('adasyn_balanced_dataset.csv')
df
```

```python
labels = ['Dengue (0)', 'Malaria (1)']
original_counts = [Counter(y)[0], Counter(y)[1]]
adasyn_counts = [Counter(y_adasyn)[0], Counter(y_adasyn)[1]]
```

```python
# Before ADASYN
plt.figure(figsize=(5, 5))
plt.bar(labels, original_counts, color=['#FF6384', '#36A2EB'])
plt.title('Class Distribution Before ADASYN')
plt.ylabel('Count')
plt.savefig('class_distribution_before_adasyn.png', dpi=400)
plt.show()
```

```python
# After ADASYN
plt.figure(figsize=(5, 5))
plt.bar(labels, adasyn_counts, color=['#FF6384', '#36A2EB'])
plt.title('Class Distribution After ADASYN')
plt.ylabel('Count')
plt.savefig('class_distribution_after_adasyn.png', dpi=400)
plt.show()
```
