```python
import pandas as pd
import matplotlib.pyplot as plt
from imblearn.over_sampling import BorderlineSMOTE
from collections import Counter
```

```python
df = pd.read_csv('encoded_dataset_28_5_25.csv')
```

```python
df.head()
```

```python
X = df.drop('CLASS', axis=1)
y = df['CLASS']
```

```python
print("Original class distribution:", Counter(y))
```

### Apply Borderline SMOTE

```python
bsmote = BorderlineSMOTE(random_state=42)
```

```python
X_bsmote, y_bsmote = bsmote.fit_resample(X, y)
```

```python
print("Borderline-SMOTE class distribution:", Counter(y_bsmote))
```

```python
X_bsmote_df = pd.DataFrame(X_bsmote, columns=X.columns)
y_bsmote_df = pd.Series(y_bsmote, name='CLASS')
balanced_df = pd.concat([X_bsmote_df, y_bsmote_df], axis=1)
```

```python
balanced_df.head()
```

```python
balanced_df.tail()
```

```python
balanced_df.to_csv('bsmote_balanced_dataset.csv', index=False)
```

```python
labels = ['Dengue (0)', 'Malaria (1)']
original_counts = [Counter(y)[0], Counter(y)[1]]
bsmote_counts = [Counter(y_bsmote)[0], Counter(y_bsmote)[1]]
```

### Before B-SMOTE

```python
plt.figure(figsize=(5, 5))
plt.bar(labels, original_counts, color=['#FF6384', '#36A2EB'])
plt.title('Class Distribution Before Borderline-SMOTE')
plt.ylabel('Count')
plt.savefig('class_distribution_before_bsmote.png', dpi=400)
plt.show()
```

```python
plt.figure(figsize=(5, 5))
plt.bar(labels, bsmote_counts, color=['#FF6384', '#36A2EB'])
plt.title('Class Distribution After Borderline-SMOTE')
plt.ylabel('Count')
plt.savefig('class_distribution_after_bsmote.png', dpi=400)
plt.show()
```
