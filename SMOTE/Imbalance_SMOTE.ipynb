```python
import pandas as pd
import matplotlib.pyplot as plt
from imblearn.over_sampling import SMOTE
from collections import Counter
```

```python
df = pd.read_csv('encoded_dataset_28_5_25.csv')
df.head()
```

```python
X = df.drop('CLASS', axis=1)
y = df['CLASS']
```

```python
X.shape
```

```python
y.shape
```

```python
print("Original class distribution:", Counter(y))
```

### Apply SMOTE

```python
smote = SMOTE(random_state=42)
X_smote, y_smote = smote.fit_resample(X, y)
```

```python
print("SMOTE class distribution:", Counter(y_smote))
```

```python
X_smote_df = pd.DataFrame(X_smote, columns=X.columns)
y_smote_df = pd.Series(y_smote, name='CLASS')
```

```python
y_smote_df
```

```python
balanced_df = pd.concat([X_smote_df, y_smote_df], axis=1)
balanced_df.to_csv('smote_balanced_dataset.csv', index=False)
```
